### Solved Question

Write a function:

common_elements(a,b)

Example:

a = [1,2,3,4]

b = [3,4,5,6]

Output:

{3,4}

```python
def common_elements(a,b):

    return set(a) & set(b)

a = [1,2,3,4]
b = [3,4,5,6]

print(common_elements(a,b))
```

Missing Numbers

Input:

nums = [1,2,4,6]

range = 1..6

Output:

{3,5}

Use set difference.

```python
def find_missing(nums):
    a = set(nums)
    d = max(a)
    b = set(range(1,d+ 1))
    return b - a

a = [1,2,4,6]

print(find_missing(a))
```

Longest Consecutive Sequence (Important)

Input:

[100,4,200,1,3,2]

Output:

4

Because sequence:

1,2,3,4

Length = 4

Hint:

Use a set for O(1) lookup.

```python
def longest_consecutive(nums):

    num_set = set(nums)
    longest = 0

    for n in num_set:

        if n - 1 not in num_set:

            length = 1
            current = n

            while current + 1 in num_set:
                current += 1
                length += 1

            longest = max(longest, length)

    return longest
```

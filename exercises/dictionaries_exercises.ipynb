### Solved Question


First Non-Repeating Element

Input:

[4,5,1,2,1,2,4]

Output:

5




````python
def first_non_repeating(nums):

    counts = {}

    for n in nums:
        counts[n] = counts.get(n, 0) + 1

    for n in nums:
        if counts[n] == 1:
            return n


a = [4,5,1,2,1,2,4]

print(first_non_repeating(a))
````

Two Sum (Dictionary Version)

Write:

two_sum(nums,target)

Example:

nums = [2,7,11,15]
target = 9


Output:

[0,1]

Constraint:

O(n)

Hint: store numbers in a dictionary.

````python
def two_sum(nums, target):

    seen = {}

    for i, num in enumerate(nums):

        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i
````

Group Words by Length

Input:

["cat","dog","apple","bat","banana"]

Output:

{

3:["cat","dog","bat"],

5:["apple"],

6:["banana"]

}


````python
def group_by(word):
    a = {}
    for i in word:

        temp = len(i)

        if temp not in a:
            a[temp] = [i]
            continue

        a[temp].append(i)
    return a




words = ["cat","dog","apple","bat","banana"]


print(group_by(words))
````

````python
# cleaner version
def group_by(words):

    groups = {}

    for word in words:

        length = len(word)

        groups.setdefault(length, []).append(word)

    return groups
````

#### String Analysis Challenge

Write a function that takes a list of strings and returns a dictionary where:
* **Keys** are the unique strings.
* **Values** are dictionaries containing:
    * `"count"`: The number of times it appears.
    * `"rank"`: Popularity rank (1 = most frequent).

**Note:** If counts are equal, rank them **alphabetically**.

**Input Example:**
```python
["apple", "banana", "apple", "cherry", "banana", "apple"]
```

**Output Example:**
```json
{
    "apple": {"count": 3, "rank": 1},
    "banana": {"count": 2, "rank": 2},
    "cherry": {"count": 1, "rank": 3}
}
```


````python
def get_frequency_rank(items):
    counts = {}
    for item in items:
        counts[item] = counts.get(item, 0) + 1

    sorted_items = sorted(counts.keys(), key=lambda x: (-counts[x], x))

    ranks = {}
    for index, item in enumerate(sorted_items):
        ranks[item] = {
            "count": counts[item],
            "rank": index + 1
        }

    return ranks

data = ["apple", "banana", "apple", "cherry", "banana", "apple"]
print(get_frequency_rank(data))
````

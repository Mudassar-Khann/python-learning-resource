Problem 1 — Student Class

Create class:

Student

Attributes:

name
age

Method:

introduce()

Output example:

Hi my name is Ali and I am 20

```python
class Student:
    def __init__(self, name , age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi my name is {self.name} and my age is {self.age}")


s1 = Student("Ali", 20)

s1.introduce()
```

Problem 2 — Rectangle

Create class:

Rectangle

Attributes:

width
height

Methods:

area()
perimeter()

Example:

Rectangle(4,5)

area → 20
perimeter → 18

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2 * (self.width + self.height)


a1 = Rectangle(4,5)

print(a1.area())
print(a1.perimeter())
```

Problem 3 — Counter Object

Create class:

Counter

Attribute:

count

Methods:

increment()

decrement()

reset()

```python
class Counter:
    def __init__(self):
        self.count = 0


    def increment(self):
        self.count += 1

    def decrement(self):
        self.count -= 1

    def reset(self):
        self.count = 0
```

Mini Project — Simple Bank Account

Create class:

BankAccount

Attributes:

owner
balance

Methods:

deposit(amount)
withdraw(amount)
display_balance()

Rules:

withdraw cannot exceed balance

```python
class AmountExceeds(Exception):
    pass
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance


    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise AmountExceeds("entered amount is more than current balance ")
        return self.balance - amount

    def display(self):
        print(self.balance)


  
```

### Solved Problems 

Write:

safe_get(d,key)

Return the value.

If key doesn't exist return:

"Key not found"

````python
def safe_get(d: dict, key):
    try:
        return d[key]
    except KeyError:
        return "key dosn't exist"
````

````python
# Cleaner Pythonic Version

# Using .get():

def safe_get(d: dict, key):
    return d.get(key, "Key doesn't exist")
````

Write:

sum_numbers(lst)

List may contain invalid elements.

Example:

[1,2,"a",3,"b",4]

Output:

10

Ignore invalid elements using try/except.

````python
def sum_numbers(lst):

    total = 0

    for i in lst:

        try:
            total += i

        except TypeError:
            pass

    return total


a = [1,2,"a",3,"b",4]
print(sum_numbers(a))


````

````python
# Even Better Design (Preferred)

# Avoid exceptions when possible.

def sum_numbers(lst):

    total = 0

    for i in lst:
        if isinstance(i, (int, float)):
            total += i

    return total
````

Write a function:

withdraw(balance, amount)

Rules:

amount < 0 → raise ValueError

amount > balance → raise ValueError

otherwise return new balance

````python
def withdraw(balance, amount):

    if amount < 0:
        raise ValueError("Amount cannot be negative")

    if amount > balance:
        raise ValueError("Insufficient funds")

    return balance - amount
````

## 1) Lists

### Quick Diagnostic Answers
- List vs Tuple: `list` is mutable, `tuple` is immutable.
- `a[0]` gives the first element.
- `a[::-1]` returns a reversed copy.
- Lists allow mixed data types and duplicates.


### Core Concept
A list is:
- Ordered
- Mutable
- Allows duplicates
- Heterogeneous (mixed types allowed)


````python
# Creating lists
a = []
b = list()
c = [1, 2, 3, 4]

# Mixed types
d = ["hi", 10, True]
print(c)
print(d)
````

````python
# Indexing and slicing
a = [0, 1, 2, 3, 4, 5]

print(a[0])    # first element
print(a[-1])   # last element
print(a[1:4])  # [1, 2, 3]
print(a[:3])   # [0, 1, 2]
print(a[::2])  # [0, 2, 4]
print(a[::-1]) # reversed copy
````

````python
# Common list methods
a = [1, 2]
a.append(3)
print(a)

a.remove(2)
print(a)

last = a.pop()
print(last, a)

b = [3, 1, 2]
b.sort()
print(b)  # sort is in-place and returns None
````

````python
# Examples
a = [5, 2, 9]
a.append(1)
a.sort()
print(a)

# Copy via slicing
b = a[:]
print(b)

# Reverse copy
print(a[::-1])
````

````python
# Example
n = ["madi", ["ka", 3], "goga", "solo", 8, 9, True ]

n[1][1] = 10
n.append(["lolopolo", 1, 4])
n.insert(0,"madigaag")

n.pop(1)
n.pop()

print(n)
print(len(n))
````

````python
# example of adding element using extend and append
a = ["mam", 2, 5, "mam"]
a.append("kisko")
a.append(["ki", 7])
# but append adds element as whole e.g in above when we appended another list instead of adding
#  individual elements it add whole list that's where extend comes into play
b = "mim"
a.extend( b)
# with extend method it adds individual elements of all iterable

print(a)
````

````python
# sorting a list
a = [2, 6, 1, 9]
a.sort()
print(a)
````

````python
# example of list storing refrence
a = [1, 2, 3, 4]
b = a
c = b
k = c
b[0] = "madi"
c.append("refrence")
print(a)
print(b)
print(c)
print(k)
# using id function to show they all pointing to same object in memory
print(id(k), id(a))
````

````python
# finding elemnt in list
a = ["mam", 2, 5, "m", "a"]
a.index("mam")
# one way is using index method it gives us the index of element
# but problem with using index is that it gives us error if element dosn't exist

print("mam" in a)
print("am" in a)
k = "am" in a
n = "mam" in a
# another method we can using membership operator that chekcs containment sequence
# it returns True if element exist and False otherwise
# you can either print it or store the boolen in a veriable to later use it in conditionals
if k:
    print("beco")
if n:
    print("beco")

print(a)
````

````python
# shallow copy
a = [1, 2, 3, 4]
b = a[:]
c = a
k = a.copy()
m = b
a.insert(4,"copy")
b[0] = "madi"
c.append("refrence")
print(a)

print(b)
print(m)
print(id(b), id(m))

print(c)
print(k)
````

### Python List Comprehension


A list comprehension creates a new list by applying an expression to every item in another iterable in a single compact line.

#### THE PROBLEM IT SOLVES

Suppose you want a list containing squares of numbers 1–5.

````python
# Without list comprehension
squares = []            # create an empty list

for i in range(1, 6):   # loop through numbers 1..5
    squares.append(i*i) # calculate square and add to list

print(squares)          # print result
````

````python
# With list comprehension
squares = [i*i for i in range(1,6)]

print(squares)
````

**Syntax**

`[expression for item in iterable if condition]`

| Part       | What it does                       | Required |
| ---------- | ---------------------------------- | -------- |
| expression | value to add to new list           | required |
| item       | variable representing each element | required |
| iterable   | source sequence                    | required |
| condition  | filter condition                   | optional |


### Extra Important Notes (Lists)
- `a = b` does not copy; both names point to the same list.
- Use `a[:]`, `list(a)`, or `a.copy()` for a shallow copy.
- For nested lists, use `copy.deepcopy()` when full independent copy is needed.
- List comprehensions are a clean way to build lists: `[x*x for x in range(5)]`.


## 2) Tuples

### Quick Diagnostic Answers
- Main difference (List vs Tuple): list mutable, tuple immutable.
- Output of:
  ```python
  t = (1, 2, 3)
  print(t[1])
  ```
  is `2`.
- Tuples are called immutable because you cannot change, add, or remove items after creation.
- `t = (5)` is not a tuple; it is an `int`. Single-item tuple must be `(5,)`.
- Tuple unpacking means assigning tuple items to multiple variables in one line.


### Core Concept
A tuple is:
- Ordered
- Immutable
- Allows duplicates
- Generally faster than list iteration
- Hashable if all elements are hashable (so it can be used as dict key)


````python
# Creating tuples
a = (1, 2, 3)
b = tuple([1, 2, 3])

single_ok = (5,)
not_tuple = (5)

print(a, b)
print(type(single_ok), type(not_tuple))
````

````python
# Indexing and immutability
t = (10, 20, 30)
print(t[0])
print(t[-1])

try:
    t[0] = 99
except TypeError:
    print("Tuple is immutable")
````

````python
# Tuple unpacking
t = (100, 200, 300)
a, b, c = t
print(a, b, c)

first, *rest = (1, 2, 3, 4, 5)
print(first, rest)
````

````python
# Useful tuple patterns
# Swap
a, b = 5, 10
a, b = b, a
print(a, b)

# Return multiple values
def pair():
    return 1, 2

x, y = pair()
print(x, y)
````

### Extra Important Notes (Tuples)
- A tuple can contain mutable objects; the tuple reference is fixed, but inner mutable values can still change.
- `namedtuple` (from `collections`) is useful when tuple positions need readable field names.
- Tuples are great for fixed records like coordinates, RGB values, and return values.

## 3) Dictionaries

### Quick Diagnostic Answers
- List vs Dictionary: list uses index positions; dictionary uses key-value mapping.
- Output of `d = {"a": 10, "b": 20}; print(d["a"])` is `10`.
- `d["x"]` raises `KeyError` if key is missing.
- Dictionary keys must be immutable and unique.
- `get()` is safe and returns `None` (or default) for missing keys; direct indexing raises error.


### Core Concept
A dictionary is:
- Mapping of key -> value
- Mutable
- Insertion-ordered in modern Python
- Fast lookup using hashing


````python
# Creating and accessing dictionaries
d1 = {}
d2 = dict()
d3 = {"a": 1, "b": 2}

d = {"name": "Alex", "age": 21, "is_student": True}
print(d["name"])
print(d.get("city"))
````

````python
data = {"a": 10, "b": 20}
# data["met"] = 10

print("a" in data)
print(20 in data.values())
print(len(data))
````

````python
# CRUD operations
d = {"a": 1}
d["b"] = 2
d["a"] = 10
print(d)

del d["b"]
print(d)
````

````python
# looping in dict
data = {"a": 10, "b": 20, "c" : 30, "d" : 40}

for i in data:
    print(i, end =" ")
print()
for i in data.values():
    print(i, end = " ")
print()
for i in data.keys():
    print(i, end=" ")
print()

for key, values in data.items():
    print(key, values)
````

#### Why Dictionaries Are Powerful

Lookups are extremely fast.

list lookup → O(n)

dict lookup → O(1)

#### Accessing Values

````python
person = {"name":"Sara","age":25}

print(person["name"])
# we can use key to access its value but it gives us key error if enterd the wrong key
````

````python
person = {"name":"Sara","age":25, "max": 90, "limit": 50}

print(person.get("name"))
a = person.get("max")
# get method also returns the value we can also store it in veriable
print(person.get("na"))
# get method dosn't gives us error incase of wrong key

print(a)
print(person)
````

#### Adding / Updating Values

````python
d = {"a":1}

d["b"] = 2
d["a"] = 10
d["kio"] = 40
print(d)
````

#### Removing Elements

````python
d = {"a":1,"b":2, "c":3, "d":4, "e":5}

d.pop("a")
# d.pop() unlike list dictionary pop dosn't pop the last key on deafult and gives error if key not enterd
m = d.pop("b")
# but as list pop method dict pop also gives return the value it poped from the given key
# we can store that value in a veriable

del d["e"]

print(d)
print(m)
````

````python
# Iteration and important methods
student = {"name": "Ali", "marks": {"math": 90, "cs": 95}}

for k, v in student.items():
    print(k, v)

print(student["marks"]["math"])
````

````python
# Example Word Counter
text = "apple banana apple mango banana apple"

counts = {}

for word in text.split():

    if word not in counts:
        counts[word] = 1
    else:
        counts[word] += 1

print(counts)
````

````python
# Example: frequency counting
text = "aabbc"
freq = {}

for ch in text:
    freq[ch] = freq.get(ch, 0) + 1

print(freq)
````

#### defaultdict


A defaultdict automatically creates a default value when a missing key is accessed, so your code doesn’t crash or require manual initialization.


````python
# Example of usage of defaultdict its a class of dicionary
#  which sets deafult value for keys if no value provided
from collections import defaultdict
a = "programming"
counts = defaultdict(int)

for i in a:
    counts[i] += 1
print(counts)
print(counts["p"])
````

````python
from collections import defaultdict

groups = defaultdict(list)

data = [("A",1),("B",2),("A",3)]

for k,v in data:
    groups[k].append(v)

print(groups)
````

#### Counter

A Counter automatically counts occurrences of elements in an iterable.

THE PROBLEM IT SOLVES


Without Counter you must manually count frequencies.

````python
from collections import Counter        # import Counter

nums = [1,2,1,3,"mam",2,1, "mam", "mam"]                   # sample list

counts = Counter(nums)                 # count occurrences

print(counts)                          # print frequency dictionary
print(counts[1])                       # frequency of 1
print(counts.most_common(1))           # most frequent element
````

````python
# Example of dict comprhention
prices_usd = {"apple":2,"banana":1}
prices_pkr = {k:v*280 for k,v in prices_usd.items()}

print(prices_pkr)
````

### Extra Important Notes (Dictionaries)
- Average-case lookup/update is O(1), but still depends on good hashing.
- Use `dict comprehension` for compact transformations:
  `{x: x*x for x in range(5)}`
- `setdefault()` can simplify grouped or nested updates.
- For frequent counting, `collections.Counter` is a strong built-in option.


## 4) Sets

### Quick Diagnostic Answers
- List vs Set: list keeps order and duplicates; set stores unique elements and is unordered.
- `s = {1,2,3,2,1}` prints `{1, 2, 3}` (order may vary).
- A set cannot contain a list because list is unhashable (mutable).
- `remove()` raises error if missing, `discard()` does not.
- `a & b` gives intersection.


### Core Concept
A set is:
- Unordered
- Mutable
- Unique elements only
- Fast membership testing


````python
# Creating sets
s1 = {1, 2, 3}
s2 = set([1, 2, 3,2])
s3 = set({1:2})  # empty set

print(s1, s2, s3)
````

````python
{}      # NOT a set (this is an empty dictionary)
set()   # correct empty set
````

````python
# Uniqueness and membership
s = {1, 1, 2, 2, 3}
print(s)
print(3 in s)
print(5 in s)
````

#### `add()`: method

````python

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

a.add(90) # Adds one element to a set if it is not already present.


b.update(9) # ignored cuse set doesn't allow duplicates elements

print(a)
print(b)
````

````python
# THE PROBLEM IT SOLVES

# Without a set you must manually check for duplicates.

users = ["ali", "sara"]          # list of users

new_user = "ali"                 # new user to add

if new_user not in users:        # check manually
    users.append(new_user)       # add if not duplicate
````

````python
# With a set:

users = {"ali", "sara"}
users.add("ali")                 # duplicate ignored automatically
````

#### `remove()`: method

Removes an element from the set but raises an error if it does not exist.

````python
numbers = {1,2,3,4}    # set

numbers.remove(3)      # remove element
# numbers.remove(5)      gives error cuse 5 dosn't exist

print(numbers)         # {1,2,4}
````

#### `discard()`: method

Removes an element if present but does nothing if it is missing.

````python
numbers = {1,2,3}

numbers.discard(3)    # removes element

numbers.discard(10)   # no error

print(numbers)
````

#### `pop()`: method

Removes and returns a random element from the set.

````python
numbers = {1,2,3,4,5,6}

value = numbers.pop()    # removes random element
v = numbers.pop()    # removes random element

print(value)
print(v)
print(numbers)
````

````python
# Example
tasks = {"task1","task2","task3"}

while tasks:
    job = tasks.pop()
    print("Processing:", job)
````

#### `clear()`: method

Removes every element from the set.

````python
numbers = {1,2,3}

numbers.clear()

print(numbers)
````

#### `update()`: method

Adds multiple elements from another iterable into the set.

````python
numbers = {1,2}

numbers.update([3,4,5])    # add multiple items
numbers.update("pro")    # add multiple items
numbers.update({"m": 1, "s": 2})    # add multiple items

print(numbers)
````

#### `copy()`: method

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

a.add(10)
d = a
c = a.copy()
a.discard(1)

print(a)
# print(b)
print(c)
print(d)
````

````python
# set union
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
c = {7, 8, 9, 10}

print(a | b | c)
print(a.union(b))
print(a.union(b).union(c))
print(a.union(b, c))
print("\n")
d = {"malika ", "misko", "diago"}
e = {"ma", "aless", "diago"}


print(d | e)
````

````python
# set intersection
a = {1, 2, 3, 4}
b = {3, 4, 2, 6}
c = {7, 2, 9, 10}

print(a & b )
print(a & b & c)
print(a.intersection(b, c))
print("\n")

d = {"malika ", "misko", "diago"}
e = {"ma", "aless", "diago"}


print(d & e)

````

````python
# set differrence
a = {1, 2, 3, 4}
b = {3, 4, 2, 6}


print(a - b)
print(b.difference(a))
````

````python
# symmetric difference
a = {1, 2, 3, 4}
b = {3, 4, 2, 6}
c = {7, 2, 9, 10}

d = {"malika ", "misko", "diago"}
e = {"ma", "aless", "diago"}


print(a ^ b)
print(a.symmetric_difference(c))
print()

print(d ^ e)

````

````python
# frozenset example (immutable set)
f = frozenset([1, 2, 3])
print(f)
````

````python
# subset and superet

A = {1,2}                 # small set
B = {1,2,3,4}             # bigger set

print(A.issubset(B))      # True → all elements of A are inside B

print(B.issuperset(A))    # True → B contains everything in A

print(B.issubset(A))      # False → B has extra elements

print(A.issuperset(B))    # False → A does not contain B
````

| Method                 | Purpose                           |
| ---------------------- | --------------------------------- |
| add()                  | add single element                |
| remove()               | remove element (error if missing) |
| discard()              | remove element safely             |
| pop()                  | remove random element             |
| clear()                | empty set                         |
| update()               | add multiple elements             |
| union()                | combine sets                      |
| intersection()         | common elements                   |
| difference()           | subtract sets                     |
| symmetric_difference() | non-common elements               |


### Extra Important Notes (Sets)
- Set membership checks are usually O(1), useful for filtering and deduplication.
- Subset/superset checks: `a <= b`, `a >= b`.
- Convert comma-separated input into clean set using `strip()`.
- Use `frozenset` when you need an immutable set (for keys or nested set-like structures).

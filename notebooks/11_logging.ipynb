### Standard Levels (in order):

| Level    | When to use it                          |
| -------- | --------------------------------------- |
| DEBUG    | Detailed internal info (for developers) |
| INFO     | General events (app running normally)   |
| WARNING  | Something unexpected but not breaking   |
| ERROR    | Something failed                        |
| CRITICAL | System is about to crash                |


```python
import logging

logging.debug("Debug info")
logging.info("App started")
logging.warning("Something odd happened")
logging.error("An error occurred")
logging.critical("System crash")
```

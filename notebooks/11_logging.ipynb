## Logging = systematic recording of events/messages during program execution

#### `Python logging Module`

Python provides a built-in module:

```python
import logging
```
What it gives you:

Structured message system

Different severity levels

Output to:

console

files

external systems

Configurable behavior

````python
import logging

logging.warning("This is a warning message")
````

### Log Levels (in order):

| Level    | When to use it                          |
| -------- | --------------------------------------- |
| DEBUG    | Detailed internal info (for developers) |
| INFO     | General events (app running normally)   |
| WARNING  | Something unexpected but not breaking   |
| ERROR    | Something failed                        |
| CRITICAL | System is about to crash                |


````python
import logging

logging.debug("Debug info")
logging.info("App started")
logging.warning("Something odd happened")
logging.error("An error occurred")
logging.critical("System crash")
````

By default, the Python logging module is configured to show messages at the
WARNING level and higher, including ERROR and CRITICAL messages. Messages at the DEBUG and INFO levels are ignored unless the logging configuration is explicitly changed.

DEBUG and INFO will NOT appear unless configured.

### Controlling Logging

````python


logging.basicConfig(level=logging.DEBUG, force=True)


logging.debug("Debug info")
logging.info("App started")
logging.warning("Something odd happened")
logging.error("An error occurred")
logging.critical("System crash")
````

````python

a = 10
b = 5

def add( num1: int, num2: int) -> int:
    return num1 + num2

def sub( num1: int, num2: int) -> float:
    return num1 - num2

def mul( num1: int, num2: int) -> int:
    return num1 * num2

def div( num1: int, num2: int) -> float:
    return num1 / num2


add_results = add(a,b)
logging.debug("{} + {} = {}".format(a, b, add_results))

sub_result = sub(a,b)
logging.debug("{} - {} = {}".format(a, b, sub_result))

mul_result = mul(a,b)
logging.debug("{} * {} = {}".format(a, b, mul_result))

divide_result = div(a,b)
logging.debug("{} / {} = {}".format(a, b, divide_result))
````

### Logging to a File

````python
import logging

logging.basicConfig(filename="info.log", level=logging.DEBUG, force=True)
# now insted of printing to consule all the info would be saved to file

def add( num1: int, num2: int) -> int:
    return num1 + num2

def sub( num1: int, num2: int) -> float:
    return num1 - num2

def mul( num1: int, num2: int) -> int:
    return num1 * num2

def div( num1: int, num2: int) -> float:
    return num1 / num2


add_results = add(a,b)
logging.debug("{} + {} = {}".format(a, b, add_results))

sub_result = sub(a,b)
logging.debug("{} - {} = {}".format(a, b, sub_result))

mul_result = mul(a,b)
logging.debug("{} * {} = {}".format(a, b, mul_result))

divide_result = div(a,b)
logging.debug("{} / {} = {}".format(a, b, divide_result))

````

### Custom Log Format

````python
import logging

logging.basicConfig(filename="info.log", level=logging.DEBUG, force=True, format="%(levelname)s - %(message)s")
# now insted of printing to consule all the info would be saved to file

def add( num1: int, num2: int) -> int:
    return num1 + num2

def sub( num1: int, num2: int) -> float:
    return num1 - num2

def mul( num1: int, num2: int) -> int:
    return num1 * num2

def div( num1: int, num2: int) -> float:
    return num1 / num2


add_results = add(a,b)
logging.debug("{} + {} = {}".format(a, b, add_results))

sub_result = sub(a,b)
logging.debug("{} - {} = {}".format(a, b, sub_result))

mul_result = mul(a,b)
logging.debug("{} * {} = {}".format(a, b, mul_result))

divide_result = div(a,b)
logging.debug("{} / {} = {}".format(a, b, divide_result))

````

# Python Debugging and Error Handling

Debugging is the process of finding and fixing problems in code. Error handling is how we prevent those problems from crashing the whole program.


## Learning Goals

By the end of this notebook, you should be able to:

- Understand tracebacks and read them correctly.
- Identify the difference between syntax errors and runtime errors.
- Use `try`, `except`, `else`, and `finally` in a clean way.
- Handle common exceptions safely and professionally.
- Apply a practical debugging workflow.


## 1. Intuition 

Programming errors happen in two common cases:

- Python cannot understand your code structure.
- Your code is valid, but something unexpected happens while it runs.

Good debugging helps you locate the failure quickly. Good error handling helps your program fail gracefully instead of crashing.


## 2. Tracebacks

A traceback is the error report Python prints when execution fails.

A traceback tells you:

- file name
- line number
- call stack (which functions were called)
- exception type and message

### How to read a traceback

Read from **bottom to top**:

- Bottom line = final error type and message.
- Lines above = where the error happened and how execution reached that point.

**Mnemonic:** Traceback = call stack + error type.


````python
def divide(a, b):
    return a / b

print(divide(10, 0))
````

## 3. Syntax Errors

A syntax error means Python cannot parse your code. This is detected **before** program execution starts.

Common causes:

- missing `:` after blocks (`if`, `for`, `while`, `def`, `class`)
- missing parentheses
- unclosed quotes or brackets
- invalid indentation

Example snippets that trigger syntax errors:

```python
if 5 > 3
    print("Hello")
```

```python
print("hello"
```


## 4. Runtime Errors (Exceptions)

Runtime errors happen while the program is executing. The code is syntactically valid, but execution hits an invalid operation.

Common exceptions:

- `ZeroDivisionError`: division by zero
- `ValueError`: invalid value for conversion or operation
- `TypeError`: incompatible data types
- `IndexError`: index out of range
- `KeyError`: dictionary key not found

## 5. `try` and `except`

Use `try` for risky code. If an exception occurs, Python immediately moves to the matching `except` block.

This allows your program to continue with a controlled message or fallback behavior.


````python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero")
````

````python
# ValueError
num_text = "abc"
try:
  int(num_text)
except ValueError:
  print("you cannot convert alphabets to int ")

# IndexError
numbers = [1, 2, 3]
try:
   print(numbers[5])
except:
  print("index out of range ")
````

````python

data = {"a":1}
try:
   print(data["b"])
except KeyError:
   print("This key dosn't exist")
````

````python
for i in range(3,0,-1):
    try:
        x = int(input("Enter the error "))
        break
    except ValueError:
        print(f"Invalid input you have {i} chances then the program will close ")
else:
    print("Your trials ended ")
````

### Catching Multiple Exceptions

Catch specific exceptions separately so each error gets a clear, relevant message.


````python
try:
    num = int(input("Enter number: "))
    result = 10 / num
except ValueError:
    print("Invalid number")
except ZeroDivisionError:
    print("Cannot divide by zero")
````

## 6. `else` and `finally`

- `else` runs only if the `try` block succeeds with no exception.
- `finally` runs **every time**, whether an error happened or not.

Use `finally` for cleanup tasks such as closing files, releasing connections, or printing final status.


````python
try:
    value = int(input("Enter a number: "))
except ValueError:
    print("Invalid input")
else:
    print("Valid number:", value)
finally:
    print("Program finished")
````

## 7. `raise` for Explicit Errors

Use `raise` when your own business rules detect invalid data.

This keeps validation logic explicit and predictable.


````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age
````

## 8. Professional Debugging Tips

1. Catch specific exceptions whenever possible.
2. Log or print the exception details during debugging:
   - `except Exception as e: print("Error:", e)`
3. Debug step by step:
   - add temporary checkpoints (`print("Step 1")`, `print("Step 2")`)
4. Keep `try` blocks small:
   - only wrap code that can fail
5. Avoid bare `except:` in normal application code:
   - it can hide real bugs

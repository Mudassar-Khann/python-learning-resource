# Object-Oriented Programming (OOP)

Object-Oriented Programming organizes code around objects that hold both data and behavior. This makes programs easier to structure, scale, and maintain.

## 1. OOP in One Sentence

Object-Oriented Programming (OOP) organizes code around objects that contain both data and behavior so programs become modular, reusable, and easier to manage.


## 2. The Problem OOP Solves

Without OOP, programs often become long collections of independent variables and functions.


````python
# Procedural style (without OOP)
student1_name = "Ali"
student1_marks = 85

student2_name = "Sara"
student2_marks = 90

def print_student(name, marks):
    print(f"{name} scored {marks}")

print_student(student1_name, student1_marks)
print_student(student2_name, student2_marks)
````

### Issues in the Procedural Approach

- Data gets scattered across many variables.
- Scaling is difficult (for 100 students, you may end up with hundreds of variables).
- Behavior is not directly connected to the data it uses.
- Reusable structure is missing.


### OOP Approach

OOP bundles related data and behavior together.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def print_student(self):
        print(f"{self.name} scored {self.marks}")

s1 = Student("Ali", 85)
s2 = Student("Sara", 90)

s1.print_student()
s2.print_student()
````

### Benefits of OOP

- Better structure
- Reusability
- Scalability
- Easier debugging
- Better real-world modeling


## 3. Core Mental Model: Blueprint to Objects

Think of a class as a blueprint and objects as real items created from that blueprint.

```text
Class (Blueprint)
      |
      v
Objects (Real instances)
```

Example mental model:

- Class: `Car`
- Attributes: `brand`, `color`, `speed`
- Methods: `start()`, `stop()`, `accelerate()`


````python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        self.speed = 0

    def start(self):
        print(f"{self.brand} is starting.")

    def accelerate(self, amount):
        self.speed += amount
        print(f"{self.brand} speed is now {self.speed} km/h")

    def stop(self):
        self.speed = 0
        print(f"{self.brand} stopped.")

car1 = Car("Toyota", "Red")
car2 = Car("Tesla", "Black")

car1.start()
car1.accelerate(30)
car2.start()
````

## 4. Classes and Objects

A class is a blueprint for creating objects, and an object is a specific instance created from that blueprint.


````python
class User:
    def __init__(self, name):
        self.name = name

user1 = User("Ali")
user2 = User("Sara")

print(user1.name)
print(user2.name)
````

## 5. Class Syntax and Naming


Professional naming convention in Python:

- Use `PascalCase` for class names.
- Examples: `UserAccount`, `BankTransaction`, `FileManager`.


## 6. Real Developer Situations

OOP is commonly used in production systems such as web apps, ecommerce backends, and games.


````python
class AppUser:
    def __init__(self, username, email):
        self.username = username
        self.email = email

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class Player:
    def __init__(self, username, health):
        self.username = username
        self.health = health

u1 = AppUser("ali123", "ali@gmail.com")
p1 = Product("Laptop", 1200)
player1 = Player("Knight", 100)

print(u1.email)
print(p1.name, p1.price)
print(player1.username)
````

## 7. Most Common Mistake

Mistake: using class attributes when each object should have its own value.


````python
# Wrong: shared class attribute for per-object data
class WrongUser:
    name = "Ali"

u1 = WrongUser()
u2 = WrongUser()

print(u1.name)
print(u2.name)
````

````python
# Correct: instance attributes are unique per object
class CorrectUser:
    def __init__(self, name):
        self.name = name

u1 = CorrectUser("Ali")
u2 = CorrectUser("Sara")

print(u1.name)
print(u2.name)
````

## 8. Quick Recap

- OOP organizes code using objects.
- Classes are blueprints; objects are instances.
- Keep data and related behavior together.
- Prefer instance attributes (`self.attribute`) for per-object values.

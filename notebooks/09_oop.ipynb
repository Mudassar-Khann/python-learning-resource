## OOP in One Sentence

Object-Oriented Programming (OOP) organizes code around objects that contain both data and behavior so programs become modular, reusable, and easier to manage.


## The Problem OOP Solves

Without OOP, programs often become long collections of independent variables and functions.


````python
# Procedural style (without OOP)
student1_name = "Ali"
student1_marks = 85

student2_name = "Sara"
student2_marks = 90

def print_student(name, marks):
    print(f"{name} scored {marks}")

print_student(student1_name, student1_marks)
print_student(student2_name, student2_marks)
````

### Issues in the Procedural Approach

- Data gets scattered across many variables.
- Scaling is difficult (for 100 students, you may end up with hundreds of variables).
- Behavior is not directly connected to the data it uses.
- Reusable structure is missing.


### OOP Approach

OOP bundles related data and behavior together.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def print_student(self):
        print(f"{self.name} scored {self.marks}")

s1 = Student("Ali", 85)
s2 = Student("Sara", 90)

s1.print_student()
s2.print_student()
````

### Benefits of OOP

- Better structure
- Reusability
- Scalability
- Easier debugging
- Better real-world modeling


## Core Mental Model: Blueprint to Objects

Think of a class as a blueprint and objects as real items created from that blueprint.

```text
Class (Blueprint)
      |
      v
Objects (Real instances)
```

Example mental model:

- Class: `Car`
- Attributes: `brand`, `color`, `speed`
- Methods: `start()`, `stop()`, `accelerate()`


````python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        self.speed = 0

    def start(self):
        print(f"{self.brand} is starting.")

    def accelerate(self, amount):
        self.speed += amount
        print(f"{self.brand} speed is now {self.speed} km/h")

    def stop(self):
        self.speed = 0
        print(f"{self.brand} stopped.")

car1 = Car("Toyota", "Red")
car2 = Car("Tesla", "Black")

car1.start()
car1.accelerate(30)
car2.start()
````

## Instance Attributes vs Class Attributes

### One Sentence

Instance attributes belong to one specific object, while class attributes are shared by all objects created from that class.

### The Problem It Solves

Programs often need two types of data:

- Data unique to each object
- Data shared by all objects

Without class attributes, you would repeat the same value in every object.

**Without class attributes (repetition)**

```python
class User:
    def __init__(self, name):
        self.name = name
        self.platform = "Twitter"

u1 = User("Ali")
u2 = User("Sara")

print(u1.platform)
print(u2.platform)
```

Problem:

- Every object stores `"Twitter"` again in memory.
- For thousands of objects this becomes inefficient.

**With class attributes (shared data)**

```python
class User:
    platform = "Twitter"

    def __init__(self, name):
        self.name = name
```

Now:

- `platform` exists once in the class.
- All objects reference it.

### Parameters

No parameters exist for defining attributes themselves.

Syntax types:

| Syntax | Meaning |
| --- | --- |
| `class_attribute = value` | Shared by all objects |
| `self.attribute = value` | Unique to one object |

### Real Developer Situations

- Website configuration: all users share the same platform name.
- Game settings: all enemies can share the same game name.


````python
class User:
    platform = "Twitter"

    def __init__(self, name):
        self.name = name

u1 = User("Ali")
u2 = User("Sara")

print(u1.name)
print(u2.name)

print(u1.platform)
print(u2.platform)
````

````python
class User:
    platform = "MyWebsite"  # class attribute: stored on the class, not inside each object

    def __init__(self, username):
        self.username = username  # instance attribute: stored in this object's __dict__

u = User("admin")

# The object stores only its own per-object data.
print(u.__dict__)

# The class stores shared values in its class namespace (__dict__).
print(User.__dict__["platform"])

# Attribute lookup through an object checks the object's __dict__ first.
# If not found there, Python checks the class.
print(u.platform)

# Changing the class attribute through the class changes the shared value.
User.platform = "NewWebsite"
print(u.platform)

# Changing it through the object does NOT change the class attribute.
# It creates a new instance attribute only for this object, which shadows the class attribute.
u.platform = "PersonalWebsite"
print(u.__dict__)
print(User.platform)
print(u.platform)

````

Python searches attributes in this order:

1. instance attributes
2. class attributes
3. parent classes

````python
class Enemy:
    game = "DarkWorld"  # shared by all Enemy objects unless an object overrides it

    def __init__(self, health):
        self.health = health

e1 = Enemy(100)
e2 = Enemy(80)

# Both objects read the same class attribute because neither object has 'game' in its own __dict__.
print(e1.game)
print(e2.game)

# Updating through the class affects all objects that still use the shared class attribute.
Enemy.game = "DarkWorld 2"
print(e1.game)
print(e2.game)

# Updating through one object creates an instance attribute only for that object.
e1.game = "CustomWorld"
print(e1.__dict__)
print(e2.__dict__)
print(Enemy.game)
print(e1.game)
print(e2.game)
````

## Methods in Classes: Instance, Class, and Static

### One Sentence

Methods are functions inside a class that operate on objects, the class itself, or neither.

### The Problem It Solves

Without methods, logic must be written outside the class, separating behavior from the data it belongs to.

**Without methods**

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

def deposit(account, amount):
    account.balance += amount
```

Problems:

- Functions live outside the object.
- Large systems become harder to organize.
- Behavior is not grouped with the data.

**With methods**

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
```

Now data and behavior live together.

### Method Types

- Instance method: works with one specific object and usually reads or changes `self`.
- Class method: works with the class itself through `cls` and is useful for shared data or alternative constructors.
- Static method: a utility function placed inside the class for logical grouping; it does not use `self` or `cls`.

### Parameters

| Method type | Main parameter | Meaning |
| --- | --- | --- |
| Instance method | `self` | The object calling the method |
| Class method | `cls` | The class itself |
| Static method | None required | No automatic object or class argument is passed |

Decorators:

- `@classmethod` is required for class methods.
- `@staticmethod` is required for static methods.


````python
class User:
    platform = "MyApp"

    def __init__(self, name):
        self.name = name

    def greet(self):  # instance method: works with one object
        print("Hello", self.name)

    @classmethod
    def get_platform(cls):  # class method: works with shared class data
        print(cls.platform)

    @staticmethod
    def info():  # static method: utility function grouped inside the class
        print("User system")

u = User("Ali")

u.greet()
User.get_platform()
User.info()
````

## Alternative Constructors with Class Methods

Sometimes programmers create alternative ways to construct an object. These are called **alternative constructors**.

In Python, they are usually implemented using `@classmethod`.

### Why Use a Class Method as a Constructor

Sometimes data comes in formats that do not match the normal constructor directly, such as:

- file data
- database records
- CSV rows
- string input
- API responses
- timestamps

Instead of forcing users to parse that data manually, the class can create objects directly.

### Basic Idea

A normal constructor expects already processed values.

```python
Student("Ali", 23)
```

But real input might arrive as one string such as:

```python
"Ali,23"
```

A class method can parse that input and then call the real constructor.

### What Happens Internally

When you write:

```python
Student.from_string("Ali,23")
```

Python effectively does this:

```python
from_string(Student, "Ali,23")
```

Inside the method, this line:

```python
return cls(name, roll)
```

becomes:

```python
Student(name, roll)
```

So the class method creates the object by calling the class itself.

### Why `cls()` Is Used

Inside a class method, `cls` means the class itself.

Using `cls(...)` instead of hardcoding `Student(...)` keeps the method correct for inheritance too. If a subclass calls the method, Python creates an object of the subclass.

### Why Not Just Use `__init__`

`__init__` is designed to receive already prepared values.

For example, this constructor expects separate integers:

```python
__init__(year, month, day)
```

But real input may look like this:

```python
"2025-03-09"
```

The class method handles parsing first, then forwards clean values into `__init__`.

### Common Professional Patterns

- `Model.from_json()`
- `User.from_dict()`
- `Config.from_file()`
- `Image.from_array()`

### Flow Visualization

```text
Date.from_string("2025-03-09")
        |
        v
parse string
        |
        v
call constructor
        |
        v
return object
```


````python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll

    @classmethod
    def from_string(cls, data):
        name, roll = data.split(",")
        return cls(name, int(roll))

s1 = Student("Ali", 23)
s2 = Student.from_string("Sara,45")

print(s1.name, s1.roll)
print(s2.name, s2.roll)
````

````python
class Date:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    @classmethod
    def from_string(cls, date_str):
        year, month, day = map(int, date_str.split("-"))
        return cls(year, month, day)

d = Date.from_string("2025-03-09")
print(d.year, d.month, d.day)

````

````python
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email

    @classmethod
    def guest(cls):
        return cls("guest", "guest@email.com")

guest_user = User.guest()
print(guest_user.username, guest_user.email)
````
